## listing data

sometimes we dont need a dataframe for storing and manipulating data, basic python data storage is suffient


you can make lists within [ ] square brackets 

```python
pressures = [0.5, 5.5, 0.234, 1.5, 2.2]
print("pressures: ", pressures)
print("length of pressures: ", len(pressures))
```

```python
print("first pressure: ", pressures[0])
```

sometimes we need to update values in the list after the fact

```python
pressures[0] = 7.8
print("new pressure is: ", pressures[0])
```

sometimes we need to add more data into our list 


```python
print("original list: ", pressures)
```

```python
# .append allows for us to add to the original list 
print("new list value: ", pressures)
```

```python
# (you can only add the end but there is a way to insert at the beginning)
# but if you want to append values anywhere excet the end

new_pressures = [1.4]
inserted_pressures = pressures[0:3] + new_pressures + pressures[3:]

print("new inserted at 4th spot: ", inserted_pressures)
```

```python
# use a "+" to combine lists 
combined_list_plus = ["A", "B"] + ["W", "X"]

# use the "extend" method to add values from one list to another
extended_list = ["A", "B"].extend(["W", "X"])
print("combined list: ", combined_list_plus)
print("extended list: ", extended_list)
```

```python
alpha_list = ["A", "B"]
alpha_list.extend(["W", "X"])
print("actually, extending: ", alpha_list)
```

list can contain a mix of diff types of data, so be careful

```python
alpha_list.append(-0.555)
alpha_list.append(["Z", 9, None])
```

```python
print(alpha_list)
```

sometimes we want to remove data from our list 

```python
del alpha_list[5]
print(alpha_list)
```

```python
text = "heres some text"
print("First character: ", text[0])
```

string variables can not be modified by index after creation, but lists can

```python
text[0] = "H"
```

# looping over data 

sometimes we need to repeat the same action multiple times on a variable but we dont want to copy and paste the code over and over again to do this

```python
data = [2, 3, 4, 5]

# this is the manual way to print all list vals
print(data[0])
print(data[1])
print(data[2])
print(data[3])

# this is how to loop print all the code
for loop_variable in data:
    print(loop_variabl)
```

```python
# loops are dependent on the tab spacing to know what code they should execute
```

```python
for loop_variable in data:
    print("printing from in the loop: ", loop_variable)
print("this is outside of the loop")
```

```python
# you can also loop using the index of the list 
```

```python
# this allows for us to see what each value is at in a list
for loop_index in range(len(data)):
    print("printing from index", loop_index)
    print("index has value: ", data[loop_index])
```

```python
# you can use loops to make your own calculations 
```

```python
sum = 0
for number in data: 
    sum = sum + number

print("total value of all list items: ", sum)
```
