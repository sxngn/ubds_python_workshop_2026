```python
def average_gdp_in_decade(country, continent, year):
    continent_data_filepath = '../data/gapminder_gdp_' + continent + '.csv'
    data_countries = pd.read_csv(continent_data_filepath, index_col=0)
    data_country = data_countries.loc[country]

    gdp_decade = 'gdpPercap_' + str(year // 10)  #this makes the string 'gdpPercap_1980'
    avg = (data_country.loc[gdp_decade + '2'] + data_country.loc[gdp_decade + '7']) / 2
    return avg 
```

```python
japan_average_gdp_decade = average_gdp_in_decade('Japan', 'asia', 1983)
print(japan_average_gdp_decade)
```
